```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Define the years

```python
import os
import re

# Define the relative path to the data directory
data_folder = "../data/extract/"
YEARS = []

# Iterate through all files in the data folder to find the years
for filename in os.listdir(data_folder):
    # Check if the file is a JPEG image
    if filename.endswith(".csv"):
        # Use regular expression to find the first sequence of digits in the filename
        match = re.search(r"\d+", filename)
        
        # If a number is found, add it to the list
        if match:
            YEARS.append(int(match.group()))

# Sort the years to maintain chronological order
YEARS.sort()

if not YEARS:
    raise ValueError("No years found in jpg files")

print("Extracted years:", YEARS)
```

```python
# df = pd.read_csv('../data/extract/2005-2023tmrland_cleaned.csv')

for year in YEARS:
    print(f"Reading: {year} lineup data")
    df = pd.read_csv(f'../data/extract/{year}_tmrland.csv')
    #df.head()
    genre_counts = df['Genre'].value_counts()
    plt.figure(figsize=(12, 6))
    plt.ylabel('Count')
    plt.xlabel('Genre')
    plt.bar(genre_counts.index, genre_counts.values, color='skyblue')
    plt.xticks(rotation=90)
    plt.yticks()
    plt.title(f'Number of Appearances by Genre ({year})')
    plt.show()
```
