# 2. Remove unecesary columns

```python
# Drop the unnecessary columns
df = df.drop(columns=['Unnamed: 0', 'Hostname', 'Stage Name', 'Timeslot', 'Date', 'Day', 'Weekend'])
df.info()
```

# 4. Break down the file into seperate years

```python
# Get unique years present in the dataframe
years = df['Year'].unique()

# Loop through each year
for year in years:
    # Filter data for the specific year
    df_year = df[df['Year'] == year]
    
    # Define the output path for this specific year
    # This will create files like: ../data/extract/2005_tmrland.csv
    output_path = f'../data/extract/{year}_tmrland.csv'
    
    # Save to CSV
    df_year.to_csv(output_path, index=False)
    print(f"Saved {year} data to {output_path}")
```
